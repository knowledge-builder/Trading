```python
# In Python, the sorted() built-in function returns a brand new sorted list from the elements of any iterable (such as lists, tuples, dictionaries, sets, or strings) without modifying the original data.

# sorted(iterable, key=None, reverse=False)

# Parameters

  # iterable: The collection you want to sort (e.g., list, tuple, string, dictionary).
  # key (Optional): A function that serves as a sorting basis (e.g., key=len to sort by length).
  # reverse (Optional): A boolean. If set to True, the list is sorted in descending order.
```

```python
# Ascending Order (Default)

numbers = [4, 1, 3, 2]
result = sorted(numbers)
print(result)   # Output: [1, 2, 3, 4]
print(numbers)  # Output: [4, 1, 3, 2] (Original list stays the same!)
```

```python
# Descending Order
# Set reverse=True to sort from highest to lowest

numbers = [4, 1, 3, 2]
result = sorted(numbers, reverse=True)
print(result)  # Output: [4, 3, 2, 1]
```

```python
# Sorting Other Iterables
# No matter what iterable type you pass in, sorted() always outputs a list

# Sorting a String
print(sorted("python"))  # Output: ['h', 'n', 'o', 'p', 't', 'y']

# Sorting a Tuple
print(sorted((5, 9, 1)))  # Output: [1, 5, 9]
```

```python
# Custom Sorting with the key Parameter

# Sorting Strings by Length
words = ["banana", "fig", "apple"]
# Sort by string length using the built-in len function
result = sorted(words, key=len)
print(result)  # Output: ['fig', 'apple', 'banana']
```

```python
# Sorting Complex Data (List of Dictionaries/Tuples)
# You can use lambda functions to point Python toward a specific item or dictionary value.

students = [
    {"name": "Alice", "grade": 88},
    {"name": "Bob", "grade": 95},
    {"name": "Charlie", "grade": 72}
]

# Sort students by their grade value
sorted_students = sorted(students, key=lambda x: x["grade"])
print(sorted_students)
# Output: [{'name': 'Charlie', 'grade': 72}, {'name': 'Alice', 'grade': 88}, {'name': 'Bob', 'grade': 95}]
```
