```python
# Higher High

import numpy as np
import pandas as pd
import mplfinance as mpf
import matplotlib.pyplot as plt

np.random.seed(42)


def generate_segment(start, end, candles=20, noise=0.3):
    x = np.linspace(start, end, candles)
    x += np.random.normal(0, noise, candles)
    return x


# Previous High
segment1 = generate_segment(100, 110)

# Pullback
segment2 = generate_segment(110, 106)

# NEW Higher High
segment3 = generate_segment(106, 116)

price = np.concatenate([segment1, segment2, segment3])

open_ = price[:-1]
close = price[1:]

high = np.maximum(open_, close) + np.random.uniform(0.3, 0.8, len(open_))
low = np.minimum(open_, close) - np.random.uniform(0.3, 0.8, len(open_))

dates = pd.date_range("2026-01-01", periods=len(open_), freq="5min")

df = pd.DataFrame(
    {
        "Open": open_,
        "High": high,
        "Low": low,
        "Close": close,
    },
    index=dates,
)

fig, axlist = mpf.plot(
    df,
    type="candle",
    style="charles",
    figsize=(14, 7),
    returnfig=True,
    title="Higher High (HH)",
)

ax = axlist[0]

idx = np.argmax(df["High"])

ax.annotate(
    "Higher High",
    xy=(idx, df["High"].iloc[idx]),
    xytext=(idx - 10, df["High"].max() + 3),
    arrowprops=dict(arrowstyle="->", lw=2),
    fontsize=12,
)

plt.show()
```

```python
# Higher Low

import numpy as np
import pandas as pd
import mplfinance as mpf
import matplotlib.pyplot as plt

np.random.seed(42)


def segment(a, b, n=20):
    return np.linspace(a, b, n) + np.random.normal(0, 0.3, n)


price = np.concatenate([
    segment(100,110),
    segment(110,104),     # Previous Low
    segment(104,118),
    segment(118,108),     # Higher Low
    segment(108,124),
])

open_=price[:-1]
close=price[1:]

high=np.maximum(open_,close)+np.random.uniform(.3,.8,len(open_))
low=np.minimum(open_,close)-np.random.uniform(.3,.8,len(open_))

dates=pd.date_range("2026",periods=len(open_),freq="5min")

df=pd.DataFrame({
    "Open":open_,
    "High":high,
    "Low":low,
    "Close":close
},index=dates)

fig,axlist=mpf.plot(
    df,
    type="candle",
    style="charles",
    figsize=(14,7),
    returnfig=True,
    title="Higher Low (HL)"
)

ax=axlist[0]

# second swing low
idx=np.argmin(df["Low"].iloc[55:])+55

ax.annotate(
    "Higher Low",
    xy=(idx,df["Low"].iloc[idx]),
    xytext=(idx+5,df["Low"].iloc[idx]-5),
    arrowprops=dict(arrowstyle="->",lw=2),
    fontsize=12
)

plt.show()
```

```python
# Lower High

import numpy as np
import pandas as pd
import mplfinance as mpf
import matplotlib.pyplot as plt

np.random.seed(42)

def seg(a,b,n=20):
    return np.linspace(a,b,n)+np.random.normal(0,.3,n)

price=np.concatenate([
    seg(100,120),
    seg(120,108),
    seg(108,116),   # Lower High
    seg(116,102)
])

open_=price[:-1]
close=price[1:]

high=np.maximum(open_,close)+np.random.uniform(.3,.8,len(open_))
low=np.minimum(open_,close)-np.random.uniform(.3,.8,len(open_))

dates=pd.date_range("2026",periods=len(open_),freq="5min")

df=pd.DataFrame({
    "Open":open_,
    "High":high,
    "Low":low,
    "Close":close
},index=dates)

fig,axlist=mpf.plot(
    df,
    type="candle",
    style="charles",
    figsize=(14,7),
    returnfig=True,
    title="Lower High (LH)"
)

ax=axlist[0]

idx=np.argmax(df["High"].iloc[35:60])+35

ax.annotate(
    "Lower High",
    xy=(idx,df["High"].iloc[idx]),
    xytext=(idx+5,df["High"].iloc[idx]+4),
    arrowprops=dict(arrowstyle="->",lw=2),
    fontsize=12
)

plt.show()
```
