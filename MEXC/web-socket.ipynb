```python
import socket

print(socket.gethostbyname("api.mexc.com"))

# ============================================================
# STEP 1 — UNDERSTANDING DNS RESOLUTION
# ============================================================

   # Code:

   # import socket

   # print(socket.gethostbyname("api.mexc.com"))


# ============================================================
# WHAT DOES THIS CODE DO?
# ============================================================

  # This converts a domain name into an IP address.

  # Example:

  # Input:

  #   api.mexc.com


  # Output:
  #   218.231.251.50

  # or:

  #     161.49.20.210

  # (The result may vary depending on DNS routing.)


# ============================================================
# WHAT IS SOCKET?
# ============================================================

# Python provides a built-in networking library called:

#     socket


# A socket is a programming interface that allows an application
# to communicate with other computers through a network.


# Simplified networking stack:


  #     Python Program

  #          |
  #          |
  #          v

  #     Operating System Network Stack

  #          |
  #          |
  #          v

  #     Internet


# The socket module gives Python access to networking functions.


# ============================================================
# WHAT IS socket.gethostbyname()?
# ============================================================

# Function:

  #     socket.gethostbyname()


  # Meaning:

  #     "Convert a domain name into an IP address."


  # Example:

  #     socket.gethostbyname("api.mexc.com")


  # Human-readable address:

  #     api.mexc.com


  # Computer-readable address:

  #     161.49.20.210


# ============================================================
# WHY DO WE NEED DNS?
# ============================================================

# Humans use domain names because they are easier to remember.


  # Humans:

  #     api.mexc.com


  # Computers:

  #     161.49.20.210


  # DNS acts as a translator:


  #     Domain Name

  #          |
  #          v

  #     IP Address


# ============================================================
# WHAT HAPPENS INTERNALLY?
# ============================================================

# When Python executes:

   #     socket.gethostbyname("api.mexc.com")


# The request follows this flow:


  #     Python Program

  #          |
  #          v

  #     socket Library

  #          |
  #          v

  #     Windows Networking Stack

  #          |
  #          v

  #     DNS Resolver

  #          |
  #          v

  #     DNS Server

  #          |
  #          v

  #     MEXC DNS Records

  #          |
  #          v

  #     IP Address Returned


# ============================================================
# RELATION TO OUR MEXC TROUBLESHOOTING
# ============================================================

# Before this, we manually tested DNS using:

  #     nslookup api.mexc.com 1.1.1.1


# Result:

  #     api.mexc.com

  #             |
  #             v

  #     a1268.r.akamai.net

  #             |
  #             v

  #     218.231.251.50


  # This confirmed:

  #     api.mexc.com

  # could successfully resolve into:

  #     218.231.251.50


# ============================================================
# RELATION TO WEBSOCKET
# ============================================================

# DNS resolution is the first step before a WebSocket connection.


# Full connection flow:


  #     Domain Name

  #          |
  #          v

  #     DNS Resolution

  #          |
  #          v

  #     IP Address

  #          |
  #          v

  #     Socket Connection

  #          |
  #          v

  #     WebSocket Handshake

  #          |
  #          v

  #     Live Market Data Stream


# ============================================================
# KEY CONCEPT
# ============================================================

# DNS does NOT connect your program to the server.

# DNS only answers:

  #     "Where is this server located?"


# Example:

# Question:

  #     Where is MEXC API?


# DNS Answer:

  #     It is located at:

  #     218.231.251.50


  # After DNS resolution succeeds:

  #     TCP
  #     TLS
  #     WebSocket

  # handle the actual communication.


# ============================================================
# SUMMARY
# ============================================================

# DNS is the internet's phonebook.

# It converts:

  #     api.mexc.com

# into:

  #     218.231.251.50


# so Python knows where to establish a network connection.

# In our MEXC architecture:

  #     DNS
  #       |
  #       v
  #     WebSocket Connection
  #       |
  #       v
  #     BTCUSDT Market Data Stream

# ============================================================
```

```python
import json
import websocket


def on_open(ws):
    print("Connected!")

    subscribe = {
        "method": "SUBSCRIPTION",
        "params": [
            "spot@public.kline.v3.api.pb@BTCUSDT@Min1"
        ]
    }

    ws.send(json.dumps(subscribe))


def on_message(ws, message):
    print(message)


def on_error(ws, error):
    print("ERROR:", error)


def on_close(ws, code, msg):
    print("Disconnected:", code, msg)


ws = websocket.WebSocketApp(
    "wss://wbs-api.mexc.com/ws",
    on_open=on_open,
    on_message=on_message,
    on_error=on_error,
    on_close=on_close,
)

ws.run_forever()
```
