```python
from faker import Faker
fake = Faker()
```

```python
import pandas as pd

ecommerce = pd.DataFrame()
```

```python
# Create customer name
customer_name = []
for _ in range(1000):
    name = fake.unique.name()
    customer_name.append(name)
   
ecommerce['customer_name'] = pd.Series(customer_name, name='customer_name')
```

```python
# Create email 
ecommerce['email'] = ecommerce['customer_name'].str.split().str[0] + '@example'
ecommerce['email']
```

```python
# Create country
countries = []
for _ in range(1000):
    country = fake.country()
    countries.append(country)

ecommerce['country'] = pd.Series(countries, name='country')
ecommerce['country']
```

```python
# Create order_id
order = []
for _ in range(1000):
    id = fake.unique.postcode()
    form = 'ORDR' + id
    order.append(form)

ecommerce['order_id'] = pd.Series(order, name='order_id') 
```

```python
# Create customer_id
customer = []
for _ in range(1000):
    id = fake.unique.postcode()
    form = 'CUST' + id
    customer.append(form)

ecommerce['customer_id'] = pd.Series(customer, name='customer_id')
```

```python
import random
import numpy as np

product_name = [
    "FreePrints 4x6 Photo Print",
    "FreePrints Square Photo Print",
    "FreePrints Glossy Photo Print",
    "FreePrints Matte Photo Print",
    "FreePrints 8x10 Photo Print",
    "FreePrints Hardcover Photobook",
    "FreePrints Softcover Photobook",
    "FreePrints Layflat Photobook",
    "FreePrints Mini Photobook",
    "FreePrints Wall Calendar",
    "Personalized Holiday Greeting Card",
    "Custom Wedding Invitation Card",
    "Birth Announcement Card",
    "Thank You Card with Photo",
    "Foil-Pressed Greeting Card",
    "Personalized Business Card",
    "Personalized Photo Mug",
    "Custom Name Necklace",
    "Personalized Photo Blanket",
    "Engraved Wooden Keepsake Box",
    "Personalized Cutting Board",
    "Custom Photo Pillow",
    "Custom Printed T-Shirt",
    "Personalized Hoodie",
    "Custom Coffee Mug",
    "Personalized Phone Case",
    "Custom Tote Bag",
    "Personalized Children’s Storybook",
    "Custom Name Puzzle Book",
    "Personalized Birthday Storybook"
]

# Create product_id
product_id = []
for _ in range(30):
    id = fake.unique.postcode()
    form = 'PROD' + id
    product_id.append(form)

product_dict = dict(zip(product_name, product_id))


# <------------ This is a probelmatic approach as the product ids and names are not evenly distributed resulting to 1 to many associations ------------> 
# creating unique product_id and map it w/ unique product names
# .map requires dictionary so we create one using dict where we use product names as keys and product_id's as values
# we create a column named ecommerce['product_id] and apply .map to assign product_id's taken from the dictionary we created

# If we are to distribute product_names and product_ids, make sure we don't break their association. In other words,
# if product_id 'PROD54712' is associated to "Personalized Birthday Storybook", this will stay intact if randomly and repeatedly distributed throughout the dataframe
# Another concern will be assigning the ids and names to their respective fields (product_id, product_name)

```

```python
# Solution:

# I just realized that I have already mapped product_names to product_ids, therefore, whatever index a product_name is located
# the associated product_id will fall on the same row

ecommerce['product_name'] = np.random.choice(product_name, size=len(ecommerce))

ecommerce['product_id'] = ecommerce['product_name'].map(product_dict)
```

```python
# Now, we need to validate if there are product_ids associated to multiple product names
# we have to avoid generating 1 to many results where 1 policy is asscoiated to multiple unique product names

validate_mapped_ids = dict(zip(ecommerce['product_id'], ecommerce['product_name']))


# Logic
# If any given product_id is associated to a product name that doesn't match what's in the dictionary, then return
# the product_id and the product_name
# boolean indexing might be the approach
# There must be a method that will check for mismatched association
# Use .get
# dict.get(key, default=None)

for k, v in validate_mapped_ids.items():
      if validate_mapped_ids.get(k) != v:
              print(f"{k}: {v}")
```
