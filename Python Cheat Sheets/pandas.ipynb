```python
df = pd.DataFrame({'A':[1,2,np.nan], 'B':[5,np.nan,np.nan], 'C':[1,2,3]})
```

```python
df
```

```python
df.dropna() # axis=0 is default and removes rows that contain NaN values
```

```python
df.dropna(axis=1) # removes columns that contain NaN values
```

```python
df.dropna(thresh=2) # thresh=int → require a minimum number of non-NA values

# You cannot use thresh and how together—they are mutually exclusive.
# If the number of non-NA values is less than thresh, the row/column is dropped.

# If you have thresh=2 and 3 columns, then:
	# Drop a row if it has 2 or more missing values (NaN).
	# Keep a row if it has fewer than 2 missing values.
```

```python
df.fillna(value=0)
```

```python
df.fillna(df.mean())
```

```python
df['A'].fillna(value=df['A'].mean())
```
