```python
# Basic error handling with try-except (intro)
import time

# Write code that asks for a number and handles the error if the input is not a number.
def input_number(input_value = input("Provide a number: ")):
    try:
        value = float(input_value)
        return value
    except ValueError:
        print(f"{input_value} is not a number")

# Handle division by zero in a program that divides two numbers.
def handle_division(input1 = int(input("Provide a number: ")), input2 = int(input("Provide a number: "))):
    try:
        result = input1 / input2
        return result
    except ZeroDivisionError:
        print("Error: Cannot divide by zero")

# Write a program that tries to access an element in a list by index and handles out-of-range errors.
def index_range(list):
    select_element = int(input("Select an element from the list: "))
    try:
        element = list[select_element]
        return element
    except IndexError:
        print("Error: index out of range")

# Write code that opens a file and handles the error if the file does not exist.
def check_file(folder):
    try:
        with open(folder, "r", encoding='utf-8') as file:
            if file.read():
                print("File found")
    except FileNotFoundError:
        print("Error: file does not exist")

# Write a program that asks for an integer and keeps asking until a valid number is entered.
def integer_input():
    while True:
        user_input = input("Please provide an integer: ")
        try:
            int_input = int(user_input)
            print(f"{int_input} is {type(int_input)}")
            break
        except ValueError:
            print(f"'{user_input}' is not a valid integer. Please try again.")

# Write a program that converts a string to int and handles invalid input.
def int_convert(str_input):
    try:
        convert_int = int(str_input)
        print(f"{convert_int} successfully converted to integer")
    except ValueError:
        print(f"{str_input} '{type(str_input)}' is not an integer string type")

# Write a program that handles multiple exceptions (ValueError and ZeroDivisionError).
def error_handling(input_value = [2, 3, 4, 5, 0, "abc"]):
    valid_integers = []
    invalid_values = []
    
    for item in input_value:
        try:
            valid_integers.append(int(item))
        except ValueError:
            invalid_values.append(item)
    
    print(f"{valid_integers} are valid integers")
    print(f"{invalid_values} invalid values")

    try:
        division = valid_integers[2] / valid_integers[4]
        print(f"Division result: {division}")
    except ZeroDivisionError:
        print("Error: divisor is invalid and must be greater or less than zero")    

# Write a program that catches exceptions and prints a friendly error message.
def error_message(CustomError):
    try:
        raise CustomError("Something is wrong")
    except CustomError as e:
        print(f"My friend there's an error: {e}")

# Write code that tries to convert input to float and handles the exception.
def float_convert(numeric_value):
    try:
        convert_value = float(numeric_value)
        print(f"{convert_value}")
    except ValueError:
        print(f"{numeric_value} '{type(numeric_value)}' is not a float type value")

# Write a program that handles keyboard interruption gracefully.
def keyboardinterrupt():
    try:
        while True:
            print("Processing...")
            time.sleep(2)
    except KeyboardInterrupt:
            print("Saving your progress...")
            print("Progress has been saved. Closing the application")
    except Exception as e:
         print(f"Unexpected error: {e}")
```

```python
# Writing and reading simple programs
from decimal import Decimal as D
import math
from typing import List

# 1. Ask for user’s name and age, then print a summary
def age_name_summary() -> None:
    name = input("Provide your name: ")
    try:
        age = int(input("What's your age: "))
    except ValueError:
        print("Invalid age. Please enter a number.")
        return
    print(f"Your name is {name}")
    print(f"You are {age} years old\n")


# 2. Convert Celsius to Fahrenheit
def convert_celsius_to_fahrenheit(celsius: float) -> None:
    F = celsius * 9 / 5 + 32
    print(f"{F}°F\n")


# 3. Calculate the area of a triangle
def calculate_area_of_triangle(base: float, height: float) -> float:
    return 0.5 * base * height


# 4. Simple calculator (+, -, *, /)
def python_calculator() -> None:
    operator = input("Choose the operator (+, -, *, /): ")
    try:
        num1 = D(input("Provide a number: "))
        num2 = D(input("Provide another number: "))
    except Exception:
        print("Invalid number input.")
        return

    try:
        if operator == '+':
            result = num1 + num2
        elif operator == '-':
            result = num1 - num2
        elif operator == '*':
            result = num1 * num2
        elif operator == '/':
            result = num1 / num2
        else:
            print("Operator is invalid. Please choose among +, -, *, /")
            return
        print(f"The result is: {result}\n")
    except ZeroDivisionError:
        print("Error: Cannot divide by 0\n")


# 5. Count vowels in a sentence
def count_vowels() -> None:
    vowels = set('aeiou')
    sentence = input("Provide a sentence: ")
    vowel_count = sum(1 for letter in sentence.lower() if letter in vowels)

    if vowel_count > 0:
        print(f"Number of vowels: {vowel_count}\n")
    else:
        print("No vowels found\n")


# 6. Reverse a list provided by the user
def reverse_list() -> List[int]:
    user_input = input("Provide numbers separated by commas: ")
    try:
        user_list = [int(x.strip()) for x in user_input.split(",")]
    except ValueError:
        print("Invalid input. Please enter only numbers separated by commas.")
        return []
    reversed_list = user_list[::-1]  # return a reversed copy
    print(f"Reversed list: {reversed_list}\n")
    return reversed_list


# 7. Generate first n Fibonacci numbers
def fibonacci_iterative(n_terms: int) -> List[int]:
    if n_terms <= 0:
        return []
    elif n_terms == 1:
        return [0]

    fib_sequence = [0, 1]
    while len(fib_sequence) < n_terms:
        fib_sequence.append(fib_sequence[-1] + fib_sequence[-2])
    return fib_sequence


# 8. Check if a number is prime
def check_prime_number() -> None:
    try:
        num = int(input("Provide a number: "))
    except ValueError:
        print("Invalid input. Please enter an integer.")
        return

    if num <= 1:
        print("Not a prime number\n")
        return

    for i in range(2, int(math.isqrt(num)) + 1):
        if num % i == 0:
            print("Not a prime number\n")
            return
    print("Prime number\n")


# 9. Store user inputs in a list until 'stop'
def stop_input() -> List[str]:
    user_list = []
    user_input = input("Provide an input (type 'stop' to exit): ")
    while user_input.lower() != "stop":
        user_list.append(user_input)
        print(f"Current list: {user_list}")
        user_input = input("Provide an input (type 'stop' to exit): ")
    print("Exiting the loop")
    print(f"Final list: {user_list}\n")
    return user_list


# 10. Write text to a file and read it back
def write_text_to_file(text_input: str, filename: str = "file.txt") -> None:
    with open(filename, "a") as f:
        f.write(text_input + "\n")
    with open(filename, "r") as f:
        print(f.read())
```
