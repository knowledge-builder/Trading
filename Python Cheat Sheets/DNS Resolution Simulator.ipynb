```python
import time
import random
import string


def rand_line(length=40):
    chars = string.ascii_uppercase + string.digits + "@#%&*"
    return ''.join(random.choice(chars) for _ in range(length))	

def rand_scan():
				chars = string.ascii_uppercase + string.digits
				return ''.join(random.choice(chars) for _ in range(25))

root_index = 0

def root_scan():
    global root_index
    roots = string.ascii_lowercase[:13]  # a–m

    result = f"contacting {roots[root_index]}.root server..."
	

    root_index = (root_index + 1) % len(roots)

    return result

print("⚙️ Starting network processes")

label_network = "[ TCP/IP stack ]" 
total_net = 20

for i in range(total_net + 1):
				percent = int((i / total_net) * 100)
				bar = "█" * i
				print(f"\r{label_network} {bar:<20} {percent:>3}%", end="", flush=True)
				time.sleep(0.08)


print("\rResult: Network stack initialized".ljust(60), end="", flush=True)

print("\n")
print("📱 Device asking for domain")
label = "> Loading" 
total = 20

for i in range(total + 1):
				percent = int((i / total) * 100)
				bar = "█" * i
				print(f"\r{label} {bar:<20} {percent:>3}%", end="", flush=True)
				time.sleep(0.08)

print("\rResult: Domain request prepared".ljust(60), end="", flush=True)

print("\n")
print("🧠 Checking local DNS cache")
for i in range(total + 1):
				percent = int((i / total) * 100)
				bar = "█" * i
				print(f"\r{label} {bar:<20} {percent:>3}%", end="", flush=True)
				time.sleep(0.05)

for _ in	range(100):
				print(f"\r> Scanning Cache {rand_scan()}".ljust(100), end="", flush=True)
				time.sleep(0.05)

print("\rResult: No match found ".ljust(60), end="", flush=True)

print("\n")
print("📡 Contacting router DNS")

for i in range(total + 1):
				percent = int((i / total) * 100)
				bar = "█" * i
				print(f"\r{label} {bar:<20} {percent:>3}%", end="", flush=True)
				time.sleep(0.05)

for _ in	range(100):
				print(f"\r> Scanning Cache {rand_scan()}".ljust(100), end="", flush=True)
				time.sleep(0.05)

print("\rResult: No cache entry — Request forwarded to ISP DNS ".ljust(100), end="", flush=True)

print("\n")
print("🌐 Contacting ISP DNS")

for i in range(total + 1):
				percent = int((i / total) * 100)
				bar = "█" * i
				print(f"\r{label} {bar:<20} {percent:>3}%", end="", flush=True)
				time.sleep(0.05)

for _ in	range(100):
				print(f"\r> Scanning Cache {rand_scan()}".ljust(100), end="", flush=True)
				time.sleep(0.05)

print("\rResult: No cache entry — Starting Recursive Lookup ".ljust(100), end="", flush=True)

print("\n")
print("🌍 Querying root DNS servers")

for i in range(total + 1):
				percent = int((i / total) * 100)
				bar = "█" * i
				print(f"\r{label} {bar:<20} {percent:>3}%", end="", flush=True)
				time.sleep(0.08)

for _ in range(20):
    print(f"\r{root_scan()}".ljust(100), end="", flush=True)
    time.sleep(0.2)

print("\rResult: .com TLD DNS servers located ".ljust(100), end="", flush=True)

print("\n")
print("🧭 Querying .com TLD DNS servers")

for i in range(total + 1):
				percent = int((i / total) * 100)
				bar = "█" * i
				print(f"\r{label} {bar:<20} {percent:>3}%", end="", flush=True)
				time.sleep(0.08)

def scan_ns(domain="google.com", delay=0.12):
    letters = string.ascii_lowercase

    for _ in range(10):
        probe = ''.join(random.choice(letters) for _ in range(3))
        print(f"\rSearching NS {probe}.{domain}".ljust(100), end="", flush=True)
        time.sleep(delay)

    print(f"\rns1.{domain}".ljust(100), end="", flush=True)
    time.sleep(0.2)
    print(f"\rns2.{domain}".ljust(100), end="", flush=True)
    time.sleep(0.2)
    print(f"\rns3.{domain}".ljust(100), end="", flush=True)
    time.sleep(0.2)
    print(f"\rns4.{domain}".ljust(100), end="", flush=True)
	
scan_ns("google.com")

print("\rResult: Authoritative DNS server for google.com located ".ljust(100), end="", flush=True)

print("\n")
print("🏢 Querying authoritative DNS server")

for i in range(total + 1):
				percent = int((i / total) * 100)
				bar = "█" * i
				print(f"\r{label} {bar:<20} {percent:>3}%", end="", flush=True)
				time.sleep(0.08)

def scan_ip(ip, delay=0.12):
    current = list(ip)
    digits = "0123456789"

    for i, ch in enumerate(ip):
        if ch.isdigit():
            for _ in range(3):
                current[i] = random.choice(digits)
                print(f"\rSearching IP {''.join(current)}".ljust(100), end="", flush=True)
                time.sleep(delay)
            current[i] = ch
            print(f"\rSearching IP {''.join(current)}".ljust(100), end="", flush=True)
            time.sleep(delay)

scan_ip("142.250.190.78")
print("\rResult: ✅ IP resolved → 142.250.190.78 ".ljust(100), end="", flush=True)
```
