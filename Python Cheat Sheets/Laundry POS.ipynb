```python
# Pricing
	# Regular wash costs ₱60 per load.
	# Heavy wash costs ₱90 per load.
	# Dry-only costs ₱50 per load.
	# Fabric softener costs ₱15 per load.
	# Extra rinse costs ₱10 per load.

class ServiceType:
    
    laundry_service = {
        "01":{"Regular": 60},
        "02":{"Heavy": 90},
        "03":{"Dry-only": 50},
        "04":{"Fabric-softener": 15},
        "05":{"Extra_rinse": 10}
				}
```

```python
from datetime import datetime

class ProcessTransaction:
	
				def __init__(self):
									
								self.service = ServiceType.laundry_service
								self.information = []
								self.summary = []
								self.calculate = []
    

				def customer_info(self):

								name = input("customer name: ")
								while True:
												phone = input("Enter phone number: ")
												try:
																phone = int(phone)
												except ValueError:
																print(f"ERROR: '{phone}' is not a valid phone number")
																continue
												break
								
								self.information.append({"name": name, "phone": phone})

								return	self.information
				

				def services_availed(self):

								while True:
											service_type = input("service id: ")
											
											if service_type	== "exit":
															break
											
											if service_type	not in self.service:
															print(f"ERROR: '{service_type}' service id not found.")
															continue

											for k, v in self.service.items():
															if k in service_type:
																					for service, price in v.items():
																									summary = {"id": k, "service": service, "price": price}
																									self.summary.append(summary)

								return	self.summary
				

				def calculate_cost(self):

								total_cost = sum(float(item["price"]) for item in self.summary)

								if total_cost	== 0:
													print("No services availed. No transaction to process.")
													return self.calculate
						
								while True:
													load = input("Enter a valid number: ")
													try:
																	load = int(load)
													except ValueError:
																	print(f"ERROR: '{load}' is not a valid number")
																	continue
													break
								if load > 2:
											final_amount = total_cost * 0.90
											discount = 0.10
								else:
											final_amount	= total_cost
											discount	= 0.00

								self.calculate.append({"final_amount": final_amount, "discount": discount, "loads": load, "total_cost": total_cost})

								return self.calculate


				def print_receipt(self):
								
								if not self.calculate:
												return
								
								print("\n=================================")
								print("--------- Laundry Receipt -------")
								print("=================================\n")

								print("Customer:")
								for i in self.information:
									print(f" • name: {i['name']}")
									print(f" • phone: +63{i['phone']}\n")

								print("Services:")
								for i in self.summary:
									print(f" • {i['service']}: ₱{i['price']}")

								print("\nTotal Cost:")
								for i in self.calculate:
									print(f" • number of loads: {i['loads']}")
									print(f" • total amount: ₱{i['total_cost']:.2f}")
									if i['discount'] > 0:
											print(f" • 10% discount applied")
									else:
											print(f" • no discount applied")
								print(f" • final amount: ₱{i['final_amount']:.2f}\n")

								print("===================================")
								print("Thank you for choosing our service!")
								current_datetime = datetime.now()
								date = current_datetime.strftime("%Y-%m-%d %H:%M:%S")
								print(date)
```

```python
transaction = ProcessTransaction()

customer = transaction.customer_info()
services	= transaction.services_availed()
calculation = transaction.calculate_cost()
receipt = transaction.print_receipt()
```
