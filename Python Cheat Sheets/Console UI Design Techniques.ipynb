```python
import random
import time

latency = random.uniform(0.2, 1.5)

print("Sending packet...")
time.sleep(latency)

print(f"Packet delivered in {latency:.2f}s")
```

```python
import string

def generate_id(length=8):
    return ''.join(random.choice(string.ascii_uppercase + string.digits) for _ in range(length))

generate_id()
```
