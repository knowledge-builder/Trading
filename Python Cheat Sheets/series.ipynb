```python
from faker import Faker
fake = Faker()
country = []

for _ in range(4):
    fake_country = fake.country()
    country.append(fake_country)
```

```python
num_ser = np.arange(1,5)

ser1 = pd.Series(num_ser, index=[v.strip("(),") for v in country])
ser1
```

```python
ser1['Fiji']
```

```python
list2 = np.arange(1,7)

ser2 = pd.Series(list2[(list2!=2) & (list2!=3)])
ser2
```

```python
ser2 = pd.Series(list2[~np.isin(list2, [2, 3])], index=ser1.index)
ser2
```
