```python
patterns = [
    r"sales_\d{4}\.csv",
    r"inventory_\d{4}\.csv"
]
```

```python
# Problem A — Validate a Sales Filename

def validate_sales_file(filename):
    
    pattern = [
        r"sales_\d{4}\.csv"
    ]

    is_valid = any(
        re.fullmatch(p, filename)
        for p in pattern
    )

    if is_valid:
        print(filename)
        print(is_valid)
    else:
        print(filename)
        print(is_valid)
       

validate_sales_file('report_2025.csv')    
```

```python
# Problem B — Support Multiple File Types

files = [
    'sales_2025.csv',
    'inventory_2024.csv',
    "sales.csv",
]

def validate_file(filename):
    valid_files = []
    invalid_files = []
    
    for f in filename:
        is_valid = any(
            re.fullmatch(pattern, f)
            for pattern in patterns
        )
        
        if is_valid:
            valid_files.append(f)
            print(f"valid file found: {f}")
        else:
            invalid_files.append(f)
            print(f"invalid file found: {f}")

validate_file(files)
```

```python
# Problem C — Count Valid Files

files = [
    "sales_2025.csv",
    "sales.csv",
    "inventory_2024.csv",
    "report.csv"
]

def count_valid_files(filename):
    valid = []
    
    for f in filename:
       is_valid = any(
        re.fullmatch(pattern, f)
        for pattern in patterns    
       )

    if is_valid:
       valid.append(f)
    
    return valid
    
count_valid_files(files)
```
